```python
import pandas as pd
import numpy as np
df=pd.read_csv("https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv")
df
```

```python
print(df.shape)
print("\n")
                  # shape, head, column names, datatypes, missing value
print(df.head)
print("\n")

print(df.columns)
print("\n")

print(df.dtypes)
print("\n")

print(df.isnull().sum())
print("\n")
```

```python
#to Check and remove duplicates
print(df.duplicated().sum())
df.drop_duplicates()
df.duplicated().sum()
```

```python
#Rename columns ("Title" > "Movie_Title")
df.rename(columns={"Title":"Movie_Title"})
```

```python
#Get unique genres
df["Genre"].unique()
```

```python
#Extract number of genres per movie
df["Genre"].str.split(",").str.len()
```

```python
#Handle missing values in 'Revenue (Millions)' and 'Metascore'

df['Revenue (Millions)'].fillna(df['Revenue (Millions)'].mean())
df['Metascore'].fillna(df['Metascore'].mean())

print(df[["Revenue (Millions)","Metascore"]].isnull().sum())
```

```python
#Top 5 movies by Rating
top5 = df.sort_values("Rating", ascending=False).head(5)

print("\nTop 5 Movies by Rating:")
print(top5[["Title", "Rating"]])
```

```python
#Movies with highest/lowest revenue
highest_revenue = df.loc[df["Revenue (Millions)"].idxmax()]
lowest_revenue = df.loc[df["Revenue (Millions)"].idxmin()]

print("\nHighest Revenue Movie:")
print(highest_revenue[["Title", "Revenue (Millions)"]])

print("\nLowest Revenue Movie:")
print(lowest_revenue[["Title", "Revenue (Millions)"]])
```

```python
#Avg. rating by Director
average_rating = df.groupby("Director")["Rating"].mean()

print("\nAverage Rating by Director:")
print(average_rating)

```

```python
#Movies per year
movies_year = df["Year"].value_counts().sort_index()

print("\nMovies per Year:")
print(movies_year)
```

```python
#Movies with rating > 8 & revenue > 100M
filtered_movies = df[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100)]
print("\n Movies with Rating > 8 and Revenue > 100:")
print(filtered_movies[["Title", "Rating", "Revenue (Millions)"]])
```

```python
#Group by Director: avg Rating, Revenue
director_group = df.groupby("Director").agg({
    "Rating": "mean",
    "Revenue (Millions)": "mean"
})

print("\nDirector Group:")
print(director_group)
```

```python
#Group by Year: count, avg Rating
year_group = df.groupby("Year").agg({
    "Title": "count",
    "Rating": "mean"
})

print("\nYear Group:")
print(year_group)
```

```python
#Create a new column Profitability:
#a) Revenue divided by Duration
#b) Find top 10 most profitable movies
df["Profitability"] = (
    df["Revenue (Millions)"] / df["Runtime (Minutes)"]
)

print("\nProfitability:")
print(df[["Title", "Profitability"]].head())
#b
top10_profit = df.sort_values(
    "Profitability",
    ascending=False
).head(10)

print("\nTop 10 Most Profitable Movies:")
print(top10_profit[["Title", "Profitability"]])
```

```python
# 17. Categorize movies by Rating

def rating_category(rating):

    if rating < 6:
        return "Poor"

    elif rating < 7.5:
        return "Average"

    else:
        return "Excellent"


df["Rating_Category"] = df["Rating"].apply(rating_category)

print("\nRating Categories:")
print(df[["Title", "Rating", "Rating_Category"]].head())
```

```python
# 18. Create a custom function to score movies

def movie_score(row):

    score = row["Rating"] * 10 + row["Revenue (Millions)"]

    return score


df["Score"] = df.apply(movie_score, axis=1)

print("\nMovie Scores:")
print(df[["Title", "Rating", "Revenue (Millions)", "Score"]].head())
```
